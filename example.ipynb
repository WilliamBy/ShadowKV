```python
import sys

# Add project root (where `models` folder is) into sys.path
PROJECT_ROOT = "/root/workspace/ShadowKV"  # TODO: change this if you move the project
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

from models.llama import Llama as LLM

llm = LLM(
    model_name="gradientai/Llama-3-8B-Instruct-Gradient-1048k",
    device="cuda:0",
    batch_size=1,
    max_length=1024 * 96,
    attn_mode="shadowkv_cpu",
    sparse_budget=1024,
    minference=True
)
```

```python
from transformers import AutoTokenizer

tk = AutoTokenizer.from_pretrained("gradientai/Llama-3-8B-Instruct-Gradient-1048k")
input_ids = tk("Hello " * 1024 * 90, return_tensors="pt")["input_ids"].to("cuda")
```

```python
output, throughput = llm.batch_generate(
    input_ids, gen_len=100, benchmark=True, temperature=0.7
)
print(f"===output===\n{output}")
print(f"===throughput===\n{throughput}")
```
